## Using a list as a queue

```python
from collections import deque

# Create a deque object using my_list
# A deque object is a double-sided queue that allows additions and removals from both ends of the queue
queue = deque(my_list)

# Append a new item to the end of the list
queue.append('Wash the car')

# Remove (pop) the item from the left side of the list
# Print the removed list item and concatenate the phrase ' - Done!'
print(queue.popleft(),' - Done!')

# Create a new list with the remaining items in the deque queue
my_list_upd=list(queue)
```

```python
# Print the new updated list
print(my_list_upd)
```

## Using a list as a stack

```python
# Using a list, create a stack of items
# A stack utilizes the LIFO (last in first out) methodology when appending and popping items 
my_list = ['Pay bills','Tidy up','Walk the dog','Go to the pharmacy','Cook dinner']
stack = []

# Append individual items from the list into the stack
for task in my_list:
    stack.append(task)

# Cycle through the items in the stack popping them one at a time
# Print the popped item with the concatenate phrase [ - Done!'
while stack:
    print(stack.pop(), ' - Done!')

# Once the stack is empty print the phase below
print('\nThe stack is empty')
```

```python
# Using the test, search for word chunks that represent potential nouns
# Print the word chunks for each noun
txt = "list is a ubiquitous data structure in the Python programming language."
doc=nlp(txt)
stk=[]
for w in doc:
    if w.pos_ == "NOUN" or w.pos_ == "PROPN":
        stk.append(w.text)
    elif (w.head.pos_ == "NOUN" or w.head.pos_ == "PROPN") and \
         (w in w.head.lefts):
        stk.append(w.text)
    elif stk:
        chunk = ''
        while stk:
            chunk = stk.pop() + ' ' + chunk
        print(chunk.strip())
```

## Removing duplicates from sequences 

```python
# Create a list
lst = ["John Silver","Tim Jemison","John Silver","Maya Smith"]

# Create a set based on the list above. Then build a list off of the set
# This will remove any duplications in the original list (lst)
lst = list(set(lst))

# Print the updated list with unique values
print(lst)
```

```python
# Perform the same process as above with one exception
# Sort the set based on the index values of the original list
lst = ["John Silver","Tim Jemison","John Silver","Maya Smith"]
lst = list(sorted(set(lst), key=lst.index))
print(lst)
```

## Performing common set operations

```python
# Create 2 sets
photo1_tags = {'coffee','breakfast','drink','table','tableware','cup','food'}
photo2_tags = {'food','dish','meat','meal','tableware','dinner','vegetable'}

# Find the values that intersect (are the same) between the 2 sets and append those values to a new set called intersection
# Then check to see if the intersection set contains more than 2 items
# If this is true, print a statement that also contains the intersection values
intersection = photo1_tags.intersection(photo2_tags)
if len(intersection)>=2:
    print(f"The photos contain similar objects:{intersection}")
```
