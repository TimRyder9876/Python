#### Re.compile()

```python
import re

pattern = re.compile(r'\d{3}-\d{2}-\d{4}')
text = "My SSN is 123-45-6789 and my friend's is 987-65-4321."
matches = pattern.findall(text)
print(matches)
```

#### Re.search()

```python
import re

result = re.search(r'World', 'Hello, World')
print(result)

result = re.search(r'Python', 'Hello, World!')
print(result)
```

#### Re.findall()

```python
import re

result = re.findall(r'\d+', 'There are 3 apples and 5 oranges.')
print(result)

result = re.findall(r'[aeiou]', 'Hello, World!')
print(result)
```

#### Re.finditer()

```python
import re

text = "The rain in Spain falls mainly in the plain."

# Changed code in book to get the 'main' portion as 'ly' is the end of the word not 'ain'
pattern = r'\b\w+ain(\w+)?\b'
for match in re.finditer(pattern, text):
    print(f"Found '{match.group()}' at position {match.start()}-{match.end()}")
```

### Working with Match Objects

#### 1. group() Method

```python
import re

pattern = r'(\w+),(\w+)'
match = re.match(pattern, 'hello,world')

print(match.group())
print(match.group(0))
print(match.group(1))
print(match.group(2))
```

#### 2. groups() Method

```python
import re

pattern = r'(\d{2})-(\d{2})-(\d{4})'
match = re.match(pattern, '12-34-5678')
print(match.groups())
```

#### 3. groupdict() Method

```python
import re

pattern = r'(?P<month>\d{2})-(?P<day>\d{2})-(?P<year>\d{4})'
match = re.match(pattern, '12-34-5678')
print(match.groupdict())
```

#### 4. start(), end(), span() Methods

```python
import re

pattern = r'\d+'
match = re.search(pattern, 'abc123def')
print(match.start())
print(match.end())
print(match.span())
```

#### Ex1: Parsing a Log File

```python
import re

log_pattern = r'(\d{4}-\d{2}-\d{2}) (\d{2}:\d{2}:\d{2}) - (.+)'
log_entry = "2023-05-15 14:30:00 - User logged in"

match = re.match(log_pattern, log_entry)

if match:
    date, time, message = match.groups()
    print(f"Date: {date}")
    print(f"Time: {time}")
    print(f"Message: {message}")
```

#### Ex2: Extracting Information from HTML

```python
import re

html = '<a href="https://www.example.com" class="link">Click here</a>'
pattern = r'<a href="(?P<url>.*?)" class="(?P<class>.*?)">(?P<text>.*?)</a>'

match = re.search(pattern, html)
if match:
    print(match.groupdict())
```

#### Ex3: Working with Multiple Matches

```python
import re

text = "The prices are $10, $20, amd $30."
pattern = r'\$(\d+)'
for match in re.finditer(pattern, text):
    price = match.group(1)
    position = match.span()
    print(f"Found price ${price} at position {position}")
```

### Using Flags

#### re.IGNORECASE (re.I)

```python
import re
text = "Python is AWESOME"

# Without IGNORECARE
print(re.search(r'python',text))

# With IGNORECASE
print(re.search(r'python', text, re.IGNORECASE))

# Using inline flag
print(re.search(r'(?i)python', text))
```

#### re.MULTILINE (re.M)

```python
import re
text = """First line
Second line
Third line"""

# Without MULTILINE
print(re.findall(r'^Second', text))

# With MULTILINE
print(re.findall(r'^Second', text, re.MULTILINE))

# With inline flag
print(re.findall(r'(?m)^Second', text))
```

#### re.DOTALL (re.S)

```python
import re

text = """First line
Second line"""

# Without DOTALL
print(re.search(r'First.*Second', text))

# With DOTALL
print(re.search(r'First.*Second', text, re.DOTALL))

# Using inline flag
print(re.search(r'(?s)First.*Second', text))
```

#### Ex1: Case-Insensitive Search in User Input

```python
user_input = input("Enter 'yes' or 'no': ").strip()
if re.match(r'^yes$', user_input, re.IGNORECASE):
    print("You said yes!")
```

#### Ex2: Processing Multi-line Configuration

```python
config = """
[Section1]
key1=value1
key2=value2

[Section2]
key3=value3
"""

sections = re.findall(r'(?m)^\[(.+)\]$', config)
print(sections)
```

#### Ex3: Matching Across Line Breaks

```python
html = """
<div>
    <p>Paragraph 1</p>
    <p>Paragraph 2</p>
</div>
"""

content = re.search(r'<div>(.*?)</div>', html, re.DOTALL)
if content:
    print(content.group(1).strip())
```
