#### Measuring Regex Performance

```python
# Using the timeit module
import timeit
import re

def test_regex(pattern, text):
    return re.search(pattern, text)

pattern = r'your pattern here'
text = 'your_test_text_here'

execution_time = timeit.timeit(lambda: test_regex(pattern, text), number=10000)
print(f"Execution time: {execution_time} seconds")

```

```python
# Using the time module for more complex scenarios
import time
import re

def measure_regex_performance(pattern, text, iterations=1000):
    regex = re.compile(pattern)

    start_time = time.time() # returns the time in seconds since Jan 1, 1970, 00:00:00(UTC)
    print(start_time)
    for _ in range(iterations):
        regex.search(text)
    end_time = time.time()
    print(end_time)
    
    total_time = end_time - start_time
    average_time = total_time / iterations

    print(f"Total time: {total_time:.6f} seconds")
    print(f"Average time per iterations: {average_time:.6f} seconds")

pattern = r"your_pattern_here"
text = "your_test_text_her"
measure_regex_performance(pattern, text)
```

#### Using Non-greedy Quantifiers

```python
# When to use non-greedy quantifiers
import re
text = '<p>First paragraph</p><p>Second paragraph</p>'

# Greedy quantifier
greedy_pattern = r'<p>.*</p>'
greedy_match = re.search(greedy_pattern, text)
print(greedy_match.group())

# Non-greedy quantifier
non_greedy_pattern = r'<p>.*?</p>'
non_greedy_match = re.search(non_greedy_pattern, text)
print(non_greedy_match.group())
```

```python
# Performance Implications
import re
import timeit

text = "a" * 10000 + "b"

def test_greedy():
    return re.search(r'a*b',text)

def test_non_greedy():
    return re.search(r'a*?b',text)

print("Greedy:", timeit.timeit(test_greedy, number=1000))
print("Non-greedy:", timeit.timeit(test_non_greedy, number=1000))
```

#### Best Practices for Writing Efficient Regex

```python
# Start specific and Gemeralize as Needed

# Start specific
pattern = r'\b[A-Za-z0-9._%+-]+@[A-Z0-9.-]+\.[A-Z]{2,}\b'

# Generalize only if needed
pattern = r'\S+@\S+\.\S+'
```

```python
# Use the right tool for the job

string = "prefix_test"

# Instead of
re.sub(r'^prefix_','',string)

# Use
string.removeprefix('prefix_')
```

```python
# Use named groups for clarity

pattern = re.compile(r'''
    (?P<year>\d{4})-
    (?P<month>\d{2})-
    (?P<day>\d{2})
''', re.VERBOSE)
```

```python
# Avoid unnecessary escapes

# unnecessary escapes
pattern = r'\[a\-z\]'

# better
pattern = r'[a-z]' 
```

```python
# Use character classes instead of alternation for single characters
pattern = r'[a-d]'
```

```python
# Optimize repeating patterns
pattern = r'\d{4}'
```

```python
# Avoid catastrophic backtracking

# potentially problematic
pattern = r'(a+)*b'

# better (if it meets your needs)
pattern = r'a*b'
```

```python
# Document your regex

pattern = re.compile(r'''
    (?P<protocol>https?://)?   # Optional protocol (http or https)
    (?P<domain>[^/]+)          # Domain name
    (?P<path>/.*)?             # Optional path
''', re.VERBOSE)
```
