```python
def evalMSE(terms,
            response,
            train,
            test):
    mm = MS(terms)
    X_train = mm.fit_transform(train)
    y_train = train[response]
    
    X_test = mm.transform(test)
    y_test = test[response]
    
    results = sm.OLS(y_train, X_train).fit()
    test_pred = results.predict(X_test)
    
    return np.mean((y_test - test_pred)**2)
```

```python
# evaluate the model using horsepower as the
# values horsepower,horsepower**2, ans horsepower**3
MSE = np.zeros(3)
for idx, degree in enumerate(range(1,4)):
    MSE[idx] = evalMSE([poly('horsepower', degree)],
                       'mpg',
                       Auto_train,
                       Auto_valid)
MSE
```

 * The MSE values for the 3 version of horsepower are 23.62, 18.76, and 18.80 respectively

## Cross Validation

```python
# CV = Auto.shape[0] causes this to perform 
# leave-one-out-cross-validation (LOOCV)
# In instead it was CV = 5, this would perform a 
# K-fold (5) cross validation

hp_model = sklearn_sm(sm.OLS,
                      MS(['horsepower']))
X, Y = Auto.drop(columns=['mpg']), Auto['mpg']
cv_results = cross_validate(hp_model,
                            X,
                            Y,
                            cv=Auto.shape[0])
cv_err = np.mean(cv_results['test_score'])
cv_err
```

```python
# Create 5 models based on an increasing power of
# horspower (1,2,3,4,5)
# store the cross validation error in cv_error

cv_error = np.zeros(5)
H = np.array(Auto['horsepower'])
M = sklearn_sm(sm.OLS)
for i, d in enumerate(range(1,6)):
    X = np.power.outer(H, np.arange(d+1))
    M_CV = cross_validate(M,
                          X,
                          Y,
                          cv=Auto.shape[0])
    cv_error[i] = np.mean(M_CV['test_score'])
cv_error
```

* There is a drop in error from horsepower and horsepower squared. Indicating horsepower squared maybe more appropriate
* Computation time of LOOCV is long

## The Bootstrap

* sklearn.ensemble.BaggingClassifier() would normally be used

```python
# Function returns an estimate for the significance level
# Normally this value is 5% instead of using something
# like this to estimate it
Portfolio = load_data('Portfolio')
def alpha_func(D, idx):
    cov_ = np.cov(D[['X','Y']].loc[idx], rowvar=False)
    return ((cov_[1,1] - cov_[0,1])/
            (cov_[0,0] + cov_[1,1]-2*cov_[0,1]))
```

```python
alpha_func(Portfolio, range(100))
```

```python
# The rng.choice is a bootstrap with replacement
rng = np.random.default_rng(0)
alpha_func(Portfolio,
           rng.choice(100,
                      100,
                      replace=True))
```

```python
def boot_SE(func, D, n=None, B=1000, seed=0):
    rng = np.random.default_rng(seed)
    first_, second_ = 0,0
    n=n or D.shape[0]
    # to loop through B times, value of loop is not used
    for _ in range(B):
        idx = rng.choice(D.index, n, replace=True)
        value = func(D, idx)
        first_ += value
        second_ += value**2
    return np.sqrt(second_ / B - (first_ / B) **2)
```

```python
# This is the the accuarcy of our estimate of the
# significance level (alpha)
alpha_SE = boot_SE(alpha_func,
                   Portfolio,
                   B=1000,
                   seed=0)
alpha_SE
```

### Estimating the Accuracy of a Linear Regression Model

```python
def boot_OLS(model_matrix, response, D, idx):
    D_ = D.loc[idx]
    Y_ = D_[response]
    X_ = clone(model_matrix).fit_transform(D_)
    return sm.OLS(Y_, X_).fit().params
```

```python
hp_func = partial(boot_OLS, MS(['horsepower']), 'mpg')
```

```python
rng = np.random.default_rng(0)
np.array([hp_func(Auto,
          rng.choice(392,392,replace=True)) for _ in range(10)])
```
