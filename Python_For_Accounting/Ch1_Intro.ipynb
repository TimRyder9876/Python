```python
# Calculate average income
import pandas as pd

data = pd.read_csv("financial_data.csv")
selected_data = data[data['Data'] == 'ABC Inc']

average_income = selected_data.Finance.mean()
print(f"The average income for ABC Inc is ${average_income:.2f}")
```

```python
# Calculate Daily Returns - Book was incorrect
import numpy as np

np.random.seed(42)
stock_prices_current = np.random.randint(50,100,10)
stock_prices_previous = np.random.randint(50,100,10)
# Calculate daily returns (this is different from what the book had)
daily_returns = (stock_prices_current - stock_prices_previous)/stock_prices_previous*100
daily_returns
```

```python
# Calculate 30 day moving average

np.random.seed(42)
account_balance = np.random.randint(70,100,30)
moving_average = np.convolve(account_balance, np.ones(30)/30, mode='valid')
print(f"Moving average of account balance: ${moving_average[0]:.2f}")
```

```python
# Perform a test of addition
import unittest

def add_numbers(a,b):
    return a+b

class TestAddition(unittest.TestCase):
    def test_add_positive_numbers(self):
        self.assertEqual(add_numbers(3,5),8)

    def test_add_negative_numbers(self):
        self.assertEqual(add_numbers(-2,-4),-6)

if __name__ == "__main__":
    unittest.main(argv=[''], verbosity=2, exit=False) # changed this from the book
```
