## Numpy

```python
import numpy as np
```

```python
# Create an array of 3 lists
jeff_salary = [2000,3000,3000]
nick_salary = [2600,2800,2800]
tom_salary = [2300, 2500, 2500]
base_salary = np.array([jeff_salary, nick_salary, tom_salary])
print(base_salary)
```

```python
# Create an array of 3 bonus lists
jeff_bonus = [500,400,400]
nick_bonus = [600,300,400]
tom_bonus = [200,500,400]
bonus = np.array([jeff_bonus, nick_bonus, tom_bonus])
print(bonus)
```

## Performing Element-wise operations

```python
# Create a new array by adding the base_salary and bonus arrays
salary_bonus = base_salary+bonus
print(salary_bonus)
```

## using Numpy Statistical Functions

```python
# Find the maximum values in the array
print(salary_bonus.max())
```

```python
# Create an array containing the maximum value in each row
print(np.amax(salary_bonus, axis=1))
```

```python
# Create an array containing the maximum value in each column
print(np.amax(salary_bonus, axis=0))
```

## Json Module

```python
import json
import pandas as pd

# Create a data dictionary in the shape of a json file
data = [
    {"Empno":9001, "Salary":3000},
    {"Empno":9002, "Salary":2800},
    {"Empno":9003, "Salary":2500}
]

# Save the dictionary as a json file
json_data = json.dumps(data)

# Read the json file into the salary dataframe
salary = pd.read_json(json_data)

# Rename the index 'Empno'
salary = salary.set_index('Empno')
print(salary)
```

```python
import pandas as pd

# Create a list of employee information
data = [['9001','Jeff Russell','sales'],
        ['9002','Jane Boorman','sales'],
        ['9003','Tom Heints','sales']]

# Create a dataframe from the data lsit
emps = pd.DataFrame(data,columns = ['Empno','Name','Job'])
column_types = {'Empno':int, 'Name':str, 'Job':str}
emps = emps.astype(column_types)
emps = emps.set_index('Empno')
print(emps)
```

## Combining Dataframes

```python
# Join the emps and salary dataframes
emps_salary = emps.join(salary)
print(emps_salary)
```

```python
# Create new series 
new_emp = pd.Series({'Name': 'John Hardy', 'Job': 'sales'}, name = 9004)

# Append the series to the emps dataframe using the name value as the index 
emps.loc[new_emp.name] = new_emp
print(emps)
```

```python
# Join the emps and salary dataframes
emps_salary = emps.join(salary)
print(emps_salary)
```

```python
# Perform an inner join on the emps and salary dataframes
emps_salary = emps.join(salary, how = 'inner')
print(emps_salary)
```

# One-to-many joins

```python
# Create order dataframe using the data list
data = [[2608,9001,35],
        [2617,9001,35],
        [2620,9001,139],
        [2621,9002,95],
        [2626,9002,218]]
orders = pd.DataFrame(data, columns=['Pono','Empno','Total'])
print(orders)
```

```python
# Merge the emps dataframe with the orders as a inner join on Empno fields
emp_orders = emps.merge(orders, 
                        how="inner", 
                        left_on="Empno",
                        right_on="Empno").set_index("Pono")
print(emp_orders)
```

```python
# Get the mean of the Empno group
print(orders.groupby(['Empno'])["Total"].mean())
```

```python
# Get the sum of the Emppno group
(orders.groupby(['Empno'])["Total"].sum())
```

# Loading the sample dataset into a panda dataframe

```python
# Load the csv file and set the columns to 'review' and 'sentiment'
# use tab as the delimiter
df = pd.read_csv('amazon_cells_labelled.txt', names=['review', 'sentiment'], sep='\t')
```

```python
# Display the first 3 rows
df.head(3)
```

# Splitting the sample dataset into training set and a test set

```python
from sklearn.model_selection import train_test_split

# Create the review (X) and sentiments (y) arrays
reviews = df['review'].values
sentiments = df['sentiment'].values

# Split the data into train and test arrays
reviews_train, reviews_test, sentiment_train, sentiment_test = train_test_split(reviews, 
                                                                                sentiments, 
                                                                                test_size=0.2, 
                                                                                random_state=500)
```

## Transforming text into numerical feature vectors

```python
from sklearn.feature_extraction.text import CountVectorizer

# Instantiate a CountVectorizer, fit the reviews(X) and transfor the ttrain and test data
vectorizer = CountVectorizer()
vectorizer.fit(reviews)
X_train = vectorizer.transform(reviews_train)
X_test = vectorizer.transform(reviews_test)
```

## Train and Evaluate the Model

```python
from sklearn.linear_model import LogisticRegression

# Create the LogisticRegression model
classifier = LogisticRegression()
classifier.fit(X_train, sentiment_train) 
```

```python
# Retrieve the accuracy score from the model
accuracy = classifier.score(X_test, sentiment_test)
print("Accuracy:", accuracy) 
```

## Making predictions on new data

```python
# Predict the sentiment of the 3 phrases using the classifier model
new_reviews = ['Old version of python useless', 'Very good effort', 'Clear and concise']
X_new = vectorizer.transform(new_reviews)
print(classifier.predict(X_new)) 
```
