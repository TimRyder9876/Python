#### Value at Risk (VaR)

```python
import pandas as pd
import numpy as np
import scipy.stats import norm

returns = pd.Series()

mu = returns.mean()
sigma = returns.std()
confidence_level = 0.95
time_horizon = 1

# Calculate VaR
z_score = norm.ppf(1-(1-confidence_level))
var = mu * time_horizon - z_score * sigma * np.sqrt(time_horizon)

print(f"Portfolio VaR at {confidence_level * 100}% confidence
        for {time_horizon}-day horizon: {var*100:.2f}%")
```
