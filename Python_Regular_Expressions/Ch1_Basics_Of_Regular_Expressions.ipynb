#### Special Characters and Meta-characters

```python
import re

text = "The quick brown fox jumps over the lazy dog. It barked 3 times."

# Match any character
pattern = r'q..k'
print(re.findall(pattern, text))

# Match word characters
# use '+' to search for words not individual letters like '\w' would do
pattern2 = r'\w+'
print(re.findall(pattern2, text))

# Match digits
pattern3 = r'\d'
print(re.findall(pattern3, text))

# Match word boundaries
pattern4 = r'\bdog\b'
print(re.findall(pattern4, text))
```

```python
import re

text = "The quick brown fox jumps over the lazy dog."

# Match words containing 'a' or 'e'
pattern1 = r'\b\w*[ae]\w*\b'
print(re.findall(pattern1, text))

# Match words starting with capital letter
pattern2 = r'\b[A-Z]\w*\b'
print(re.findall(pattern2, text))

# Match words not containing 'e'
pattern3 = r'\b\w*[^e\s]+\b'
print(re.findall(pattern3, text))
```

#### Quantifiers: *, +, ?, {n,m}

```python
import re

text = "The quick brown fox jumps over the lazy dog. The dog's name is Fido and he is very friendly."

# Match words with 3 or more characters
pattern1 = r'\b\w{3,}\b'
print(re.findall(pattern1, text))

# Match words starting with 'f', case insensitive
pattern2 = r'\bf\w*\b'
print(re.findall(pattern2, text, re.IGNORECASE))

# Match 'o' followed by 0 or more 'x'
pattern3 = r'ox*'
print(re.findall(pattern3, text))

# Match 'o' followed by 1 or more 'v'
pattern4 = r'ov+'
print(re.findall(pattern4, text))

# Match 'do' optionally followed by 'g'
pattern5 = r'dog?'
print(re.findall(pattern5, text))
```

#### Escaping Special Characters

#### Grouping and Capturing

```python
# Grouping
# Grouping using () allows you to apply quantifiers to a group of characters

import re
text = "hahaha hehehe hohoho"

# Match the 'ha' repeated two or more times
pattern1 = r'(ha){2,}'
print(re.findall(pattern1, text))

# Match any two-letter around repeated two or more times
pattern2 = r'([a-z]{2})\1+'
print(re.findall(pattern2, text))
```

```python
# Capturing
# When you use () to group parts of a regular expression, those gruops
# are autimatically "captured". This mean you can retrieve the part
# of the string that matched each group

import re
text = "John Smith: 1234567890"

pattern = r'(\w+)\s(\w+):\s(\d+)'
match = re.search(pattern,text)

if match:
    print(f"Full Name: {match.group(1)} {match.group(2)}")
    print(f"Phone Number: {match.group(3)}")
```

```python
# Non-capturing groups
pattern = r'(?:\w+)\s(\w+)'
match = re.search(pattern, text)
print(match.group(1))
```

#### Lookahead and Lookbehind Attributes

```python
# Positive Lookahead

import re
text = "I have 20 apples and 30 oranges"

# Match numbers followed by 'apples'
pattern = r'\d+(?=\s*apples)'
print(re.findall(pattern, text))
```

```python
# Negative Lookahead

# Match numbers not followed by 'apples'
# Book code does not align to expected result
pattern = r'\d+(?!\sapples)'
print(re.findall(pattern, text))
```

```python
# Positive Lookbehind

# Match 'apples' preceded by a number
# Code changed ue to fixed width issue when using a positive lookbehind
pattern = r'(?<=\d{2}\s)apples'
print(re.findall(pattern, text))
```

```python
# Negative Lookbehind

# Match 'oranges' not preceded by '20'
pattern = r'(?<!20\s)oranges'
print(re.findall(pattern, text))
```

#### Flags and Options

```python
import re

text = """The quick brown fox
Jumps over the lazy dog.
The dog's name is Fido
And he is very friendly."""

# Case-insensitive matching
pattern1 = r'the'
print(re.findall(pattern1, text, re.IGNORECASE))

# Multiline matching
pattern2 = r"^The.*"
print(re.findall(pattern2, text, re.MULTILINE))

# Dot matching newline
pattern = r'fox.*dog'
print(re.findall(pattern, text, re.DOTALL))

# Verbose mode
pattern4 = re.compile(r"""
    \b    # Word boundary
    \w+   # One or more word characters
    z\w+  # 'z' followed by one or more word characters
    \b    # Word boundary
""", re.VERBOSE)
print(pattern4.findall(text))
```
