#### Audit Trail

```python
import datetime

def log_change(name,action,amount):
    timestamp = datetime.datetime.now()
    log_entry = f"{timestamp} - {name} {action} ${amount}"

    # Append the log entry to a file
    with open("audit_log.txt", "a") as log_file:
        log_file.write(log_entry +"\n")

# Simulate a transaction
name="John"
action = "added"
amount = 1000
log_change(name,action,amount)
```
