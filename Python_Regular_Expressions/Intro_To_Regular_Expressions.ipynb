```python
import re
```

```python
# 02 Special Characters (Metacharacters
# 'r' for raw string
# '^; for start of the string
# '.' matches any character except a newline
# '*' match 0 or more repetitions of the previous character or group
# '$' matches the end of a string

# Below matches strings starting with "py" and ending with "n"
# Any number of characters can be in betweeen '.*'
pattern = r"^py.*n$"
```

```python
# 05 Quantifiers
# 'r' for raw string
# '\d' for digits (0-9)
# '{3}' 3 digits for occur

# this is a social security number pattern
pattern = r"\d{3}-\d{2}-\d{4}"
```

```python
# 06 Grouping and Capturing
#'r' for raw string
# '()' for grouping of characters
# '\w' for any alphanumeric or underscore character
# '+' for 1 or more times (/w characters in this case)
# '\s' for whitespace character (space, tab, new line)
# '\1' is a back reference to the first captured grouping '()'
#      in this case it is looking for a repeat of the group

# Below is a pattern to look for 2 words repeated next to each other
pattern = r"(\w+)\s+\1"
```

```python
# 07 Alternation
# '|' the pipe symbol is for an 'OR' 

pattern = "cat|dog"
```

```python
# 09 Lookahead and Lookbehind Assertions
# 'r' for raw string
# '\b' (x2) for word boundary
# '\w' for any alphanumeric or underscore character
# '+' for 1 or more times
# '(?= )' for positive look ahead
# '\b' for end of word boundary

# match words ending in 'ing'
pattern = r"\b\w+(?=ing)\b"
```

```python
# 11 Raw Strings
# use r before string for raw strings
# prevents Python from interpreting backslashes in the pattern

#\b is a word boundary
pattern = r"\bword\b"
```

```python
# 12 Non-Capturing Groups
# 'r for raw string
# '(?:...)' to signify that the contents of the group will not be used for later retrieval
#            '(?:http?://)'
# 's?' to signify the 's' may or may not be part of the matching string (https OR http) 
# '?' to signify the 'http(s)' may or may not exist in the match
# '(?:...)' to signify that the contents of the group will not be used for later retrieval
#            '(?:www\.)'
# '\.' used to make the '.' a literal value and not a '.' which is for matching any character
# '?' to signify the 'www.' is optional in the match
# '\.' used to make the '.' a literal value and not a '.' which is for matching any character

# searches for the phrase 'example.com' that coild be a part of 'http(s)://' and/or 'www.'
pattern = r"(?:https?://)?(?:www\.)?example\.com"
```

```python
# 13 Named Groups
# 'r for raw string
# '(?P<year>...)' to signify a named capturing group called year
# '\d{4}' to signify 4 digits
# '(?P<month>...)' to signify a named capturing group called month
# '\d{2}' to signify 2 digits
# '(?P<day>...)' to signify a named capturing group called day
# '\d{2}' to signify 2 digits

# search for yyyy-mm-dd and break into year, month, and day groups
pattern = r"(?P<year>\d(4) - (?P<month>\d{2}) - (?P<day>\d{2})"
```

#### Common Regex Patterns in Python

```python
# 01 Email addresses
# '\b' is a word boundary
# [A-Za-z0-9._%+-]+ for one or more characters, dots, or special characters
# '\.' to signify the literal '.'
# [A-Za-z]{2,} for 2 or more letters

pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"
```

```python
email = "user.name_tag@example.com"
if re.match(pattern, email):
    print("Valid email")
```

```python
# 02 URLs

# 's? to signify the  may or may not exist in the match
# '(?:www\.)?' to signify the 'www.' is optional and will not be used for later retrieval
# '[\w.-]+' to signify a class of any alphanumeric or '.' or '-' character one or more times
# '\.' to signify a literal '.' character
# '[a-zA-Z]{2,}' any letter 2 or more times
# '(?:/\S*)' to signify an optional '/' followed by 1 or more non-whitespace values (an optional directory path on a URL)

pattern = r"https?://(?:www\.)?[\w.-]+\.[a-zA-Z]{2,}(?:/\S*)?"
```

url = "https://www.example.com/path/to/page"

```python
# 03 Phone Numbers
# '\(?'     = a literal '(' value that is optional from the '?'
# '(\d{3})' = a 3 digit value in a capturing group
# '\)?'     = a literal ')' value that is optional from the '?'
# '[-.\s]?' = an optional '-' or '.' or whitespace character
# '(\d{3})' = a 3 digit value in a capturing group
# '[-.\s]?' = an optional '-' or '.' or whitespace character
# '(\d{4})' = a 4 digit value in ia capturing group 

pattern = r"\(?(\d{3})\)?[-.\s]?(\d{3})[-.\s]?(\d{4})"
```

```python
phone = "(123) 456-7890"
match = re.search(pattern, phone)
if match:
    print(f"Area code: {match.group(1)}")
    print(f"Phone number: {match.group(2)}-{match.group(3)}")
```

```python
# 04 Dates

# For month value
# (...)       = for capturing group
# '0[1-9]'    = starts with 0 and then another digit 1 to 9
# '|'         = OR logic
# '1[0-2]'    = starts with 1 and then another digit 0 to 2
# '|'         = OR logic

# For day value
# (...)       = for capturing group
# '0[1-9]'    = starts with 0 and then another digit 1 to 9
# '|'         = OR logic
# '[12][0-9]' = starts with 1 or 3 and then another digit 0 to 9
# '|'         = OR logic
# '3[01]      = starts with3 and then another digit 0 or 1

# For year value
# (...)       = for capturing group
# '\d{4}'     = 4 digits
pattern = r"(0[1-9]|1[0-2])/(0[1-9]|[12][0-9]|3[01])/(\d{4})"
```

```python
date = "12/31/2022"
match = re.match(pattern, date)
if match:
    month, day, year = match.groups()
    print(f"Year: {year}")
    print(f"Month: {month}")
    print(f"Day: {day}")
```

```python
# 05 IP Addresses
# '/b'        = word boundary
# (?:...)     = non-capturing group
# '25[0-5]'   = match number between 250 to 255
# |           = OR
# 2[0-4][0-9] = match number between 200 to 249 
# |           = OR
# [01]?       = matches optional number of 0 or 1
# [0-9]       = matches number between 0 and 9
# [0-9]?      = matches optional number between 0 and 9
# '\.'        = literal '.' dot
# {3}         = the preceding group should be repeated exactly 3 times
# (?:25[0-5] | 2[0-4][0-9] | [01]?[[0-9][0-9]?)    = repeat the same 0 to 255 pattern from above

# any spaces in the expression are considered part of the matching process

pattern = r"\b(?:(?:25[0-5]|2[0-4][0-9]|[01]?[0-9][0-9]?)\.){3}(?:25[0-5]|2[0-4][0-9]|[01]?[0-9][0-9]?)\b"
```

```python
ip = "152.168.0.1"
if re.match(pattern, ip):
    print("Valid IP Address")
else:
    print("Invalid IP address")
```

```python
# 06 HTML tags
# '<'      = match the literal value '<'
# [^>]+    = match any value 1 or more times except the '>' character
# '>'      = match the literal value '>'

pattern = r"<[^>]+>"
```

```python
# 07 Password Strength
# '^'                    = asserts the beginning of the string
# '(?=.*[a-z])'          = look forward into the string to have one or more a-z values
# '(?=.*[A-Z])'          = look forward into the string to have one or more A-Z values
# '(/=.*\d)'             = look forward into the string to have one or more digits
# '(?=.*[@$!%*?&])'      = look forward into the string to have one or more special characteres
# [A-Za-z\d@$!%*?&]{8,}  = allowed character list upper, lower case, digits, special characters
# {8,}                   = 8 minimum characters

pattern = r"^(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@$!%*?&])[A-Za-z\d@$!%*?&]{8,}$"
```

```python
password = "StrongP@ss1"
if re.match(pattern, password):
    print("Strong password")
```

```python
# 09 Social Security Numbers
# '\b'                       = word boundary
# '(?!000|666|9\d{2})\d{3}   = 3 digits but not 000, 666, or an 9XXs
# '(?!00)\d{2}               = 2 digits but not 00
# '(?!0000)\d{4}'            = 4 digits but not 0000

pattern = r"\b(?!000|666|9\d{2})\d{3}-(?!00)\d{2}-(?!0000)\d{4}\b"
```

```python
ssn = '123-45-0001'
if re.match(pattern, ssn):
    print("Valid SSN format")
else: print("Invalid SSN format")
```

```python
# 10 ZIP Codes
# '\b'     = word boundary
# '\d{5}'  = 5 digits
# (?:-\d{4})  = optional '-' with 4 digits

pattern = r"\b\d{5}(?:-\d{4})?\b"
```

```python
# 11 Time in 24-hr Format
# '[01]\d'  = matches the hours from 00 to 19
# '|'       = OR
# '2[0-3]'  = matches the hours from 20 to 23
# ':'       = matches the ':' character
# ([0-5]\d) = matches the minutes from 00 to 59
# '$'       = the anchor that matches theend of the string

pattern = r"^([01]\d|2[0-3]):([0-5]\d)$"
```

```python
# 19 Extracting YouTube Video Ids
# '(?:...)'             = non-capturing group
# 'https?:\/\/'         = http:// or https:// match
# '()?'                 = optional non-capturing group for (http(s))
# '(?:www\.)?           = optional non-capturing group for 'www.' phrase
# (?:youtube\.com)      = non-capturing group for 'youtube.com'
# '|'                   = OR
# 'youtu\.be'           = match 'youtu.be'
# '\/'                  = match '/'
# '(?:watch\?v=)?'      = optional non-capturing group 'watch?v='
# '(?:embed\/)?'        = optional non-capturing group 'embed\'
# '(?:shorts\/)?'       = optional non-capturing group 'shorts\'
# '(?:live\/)?'         = optional non-capturing group 'live\'
# '(?:(?!videos)...)?'  = optional non-capturing group negative lookahead ensuring 'video' is not next
# '(?!\?)'              = optional negative lookahead ensuring '?' is not next
# '(... .+\/)?'         = optional one or more characters followed by a '/'
# '(.{11})'             = match 11 characters that targes the video ID

pattern = r'(?:https?:\/\/)?(?:www\.)?(?:youtube\.com|youtu\.be)\/(?:watch\?v=)?(?:embed\/)?(?:v\/)?(?:shorts\/)?(?:live\/)?(?:(?!videos)(?!\?).+\/)?(.{11})'
```

```python
url = "https://www.youtube.com/watch?v=dQw4w9WgXcQ"
match = re.search(pattern, url)
if match:
    print(f"Video ID: {match.group(1)}")
```

```python
# 20 Matching Time Zones
# (...)?         = capturing group
# (?:GMT|UTC)    = non-capturing group matching 'GMT' OR 'UTC'
# (?:[+-]...)?   = optional non-capturing group matching '-' or '+' characters
# '\d{1,2}'      = matches 1 or 2 digits
# '(?::\d{2})?'  = optional non-capturing matching ':' and 2 digits

pattern = r'((?:GMT|UTC)(?:[+-]\d{1,2}(?::\d{2})?)?)'
```

```python
text = "The meeting is scheduled for 3 PM GMT+2"
match = re.search(pattern, text)
if match:
    print(f"Time zone: {match.group(1)}")
```

```python
# 21 Extracting Version Numbers
# '(...)'      = create a capturing group
# '\d+'        = matching 1 or more digits
# '\.'         = matches the '.' dot
# '(?:\d+\.)'  = non-capturing group matching 1 or more digits followed by a '.' dot
# '*'          = matches zero or more occurences of the preceeding line (digits + '.')
# '\d+'        = matching 1 or more digits

pattern = r'(\d+\.(?:\d+\.)*\d+)'
```

```python
text = "Python version 3.9.5 was released in May 2021"
version = re.search(pattern, text)
if version:
    print(f"Version: {version.group(1)}")
```

#### Advanced Regex Techniques in Python

```python
# 01 Lookahead and Lookbehind Assertions
# '\b'     = word boundary
# '\w+'    = one or more characters
# (?=...)  = creates a positive look forward pattern
# '\s+'    = one or more whitesapces
# '\d+'    = one or more digits

# return the word(s) in a word - whitespace - number pairings
pattern = r'\b\w+(?=\s+\d+)'
```

```python
text = "apple 123 banana 456"
matches = re.findall(pattern, text)
print(matches)
```

```python
# 02 Non-Capturing Groups
# (?:...)    = creates a non-capturing group
# 'Mr\.|Mrs\.'  = matches 'Mr.' OR 'Mrs'
# '\s+\         = matches one or more whitepaces
# '(\w+)'       = creates a capturing group of one of more alphanumeric characters

# Match "Mr." or "Mrs." followed by a name but only capture the name
pattern = r'(?:Mr\.|Mrs\.)\s+(\w+)'
```

```python
text = "Mr. Smith and Mrs. Johnson"
matches = re.findall(pattern, text)
print(matches)
```

```python
# 03 Named Groups
# '(?P<name> ...) creates a named group

# match yyyy-mm-dd date
pattern = r"(?P<year>\d{4})-(?P<month>\d{2})-(?P<day>\d{2})"
```

```python
date = "2023-05-15"
match = re.match(pattern, date)
if match:
    print(match.groupdict())
```

```python
# 04 Conditional Patterns
# '(\()?'     = create optional capturing group for the '(' character
# '(\d+)'     = create a capturing group that matches one or more digits
# (?(1)\))'   = creates a conditional matching group on the first capturing group if it exists
#               whereby the match is ')' but only if the first group exists

pattern = r'(\()?(\d+)(?(1)\))'
```

```python
text = "123 (456) 789"
matches = re.findall(pattern, text)
print(matches)
```

```python
# 05 Recursive Pattern
# Using matching nested parenthesis to simulate recursion
# '\('      = matches '(' character
# '(?:...)' = creates nested non-capturing group 
    # '[^()]*'   = matches one or more characters except '(' and ')'
    # '|'        = OR
    # '\('       = matches the '(' character
    # '(?:...)*  = creates non-capturing group 
        # '[^()]'   = matches one or more characters except '(' and ')'
        # '|'       = OR
        # '\('      = matches the '(' character    
        # '[^()]*'  = matches one or more characters except '(' and ')'
        # '/)'      = matches the '(' character
    # '/)'       = matches the '(' character
# '/)'       = matches the '(' character

pattern = r'\((?:[^()]|\((?:[^()]|\([^()]*\))*\))*\)'
```

```python
text = "test ((a+b)*(c-d)) test"
matches = re.findall(pattern, text)
print(matches)
```

```python
# 06 Atomic Grouping
# An atomic group, once a match is found for its internal pattern, 
# discards all backtracking information for that group.

# 'a'     = matches 'a'
# (?>...) = this is an atomic group
    # '.*'   = matches zero or more single characters except a newline character 
# 'b'     = matches 'b'/

pattern = r'a(?>.*)b'
```

```python
# 13 Using Regex with File Operations
# Replaces the word 'old' with the word 'new'

def replace_in_file(filename, pattern, replacement):
    with open(filename, 'r+') as file:
        content = file.read()
        content_new = re.sub(pattern, replacement, content)
        file.seek(0)
        file.write(content_new)
        file.truncate()

replace_in_file('example.txt', r'\bold\b', 'new')
```

#### Common Regex Pitfalls and How to Avoid Them

```python
# 01 Catastrophic Backtracking
# certain patterns can cause exponential time complexity on non-matching inputs

pattern = r'(a+)+b'  # bad pattern
pattern = r'a+b'     # better pattern
```

```python
# 02 Greedy vs Lazy Quantifiers
# Greedy quantifiers can match more than intended

text = '<div>content</div><div>more</div>'
pattern = r'<div>.*</div>'  # bad pattern
pattern = r'<div>.*?</div>' # good pattern
```

```python
# 03 Overuse of Capturing Groups
# Unnecessary capturing groups can slow down regex and complicate results
# Use non-capturing groups when you don't need to extract that part

pattern = r'(\d+)-(\d+)-(\d+)'     # bad pattern
pattern = r'(?:\d+)-(?:\d+)-(\d+)' # better pattern
```

```python
# 04 Not Escaping Special Characters
# Special regex characters can change the meaning of your pattern

pattern = r'1+1=2'  # '+' is treated as a quantifier and not the literal '+' plus sign
pattern = r'1\+1=2' # better pattern
```

```python
# 05 Incorrect Character Classes
# Misunderstanding how character classes work can lead to incorrect matches
pattern = r'[a-Z]'    # This will not match all letters
pattern = r'[a-zA-Z]' # better pattern
```

```python
# 06 Overreliancs on '.*'
# Using .* can lead to unexpected matches and performance issues
pattern = r'start.*end' # should be more specific what can appear between start and end

# '[^]'    = the empty character set often used to match any character including newline
# '*?'     = non-greedy quantifier signifying matching as few times as possible any
#            characters including newline  
pattern = r'start[^]*?end'     # better pattern


# '[\s\S]' = matches any whitespace or non-whitespace characters including newline
# '*?'     = non-greedy quantifier signifying matching as few times as possible any
#            characters including newline 
pattern = r'start[\s\S]*?end'  # better pattern
```

```python
# 07 Incorrect Anchoring
# Not using anchors can lead to partial matches where full matches are intended

pattern = r'\d+'  # Matches any sequence of digits, even within larger strings

# '^'   = match MUST occur at the beginning of a string
# '\d+' = match one or more digits
# '$'   = matches the end of the string
pattern = r'^\d+$'  # better pattern

# '\b...\b' = word boundary anchor
pattern = r'\b\d+\b'  # better pattern
```

```python
# 08 Misunderstanding Whitespace Handling
# Forgetting that \s matches all whitespace, including newlines

pattern = r'start\s+end' # Might match across lines
pattern = r'start[ \t]end'  # the class is used to match only whitespaces and tabs '\t' 
                            # better pattern                             
```

```python
# 09 Forgetting to Use Raw Strings
# Escape sequences in regular strings can interfere with regex syntax

#pattern = '\d+\.\d+'
pattern = r'\d+\.\d+'  # better pattern
```

```python
# 10 Overlooking Unicode
# Pitfall: Assuming \w, \d, etc. only match ASCII characters

pattern = r'w+'  # might not match non-ASCII letters
pattern = re.compile(r'\w+', re.UNICODE)  # better pattern
```

```python
# 11 Not Considering Word Boundaries
# Matching substrings when whole words are intended

pattern = r'cat' # matches 'category' as well
pattern = r'\bcat\b'  # better pattern
```

```python
# 12 Inefficient Alternation
# Using alernation | inefficiently can lead to slower matching

pattern = r'apple|banana|orange|grape'

# order alternatives from most to least specific, and use character
# classes for single-character alternations
pattern = r'banana|orange|apple|grape' # 'banana is more specific than 'apple'
```

```python
# 13 Not Accounting for Case Sensitivity
# Regex is case-sensitive by default

pattern = r'python'  # will not match 'Python'
pattern = re.compile(r'python', re.IGNORECASE)
```

```python
# 14 Overcomplicating Patterns
# Writing overly complex patterns that are hard to read and maintain

pattern = r'^(?=.*[A-Z])(?=.*[a-z)(?=.*\d)(?=.*[!@$%&?*])[A-Za-z\d@$!%*?&]{8,}$'
pattern = re.compile(r"""
    ^                      # Start of the string
    (?=.*[A-Z])            # At least one uppercase letter
    (?=.*[a-z])            # At least one lowercase letter
    (?=.*\d)               # At least one digit
    (?=.*[!@$%&?*])        # At least one special character
    [A-Za-z\d@$!%*?&]{8,}  # At least 8 characters long
    $                      # End of the string
""", re.VERBOSE)
```
