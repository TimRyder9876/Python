#### Grouping

```python
import re

text = "abababab"
pattern = r"(ab)+"
match = re.match(pattern, text)
print(match.group())
```

#### Capturing

```python
import re

pattern = r"(\d{2})-(\d{2})-(\d{4})"
text = "Date: 15-03-2024"
match = re.search(pattern, text)
if match:
    print(f"Day: {match.group(1)}")
    print(f"Month: {match.group(2)}")
    print(f"Year: {match.group(3)}")
```

#### Name Groups

```python
import re

pattern = r"(?P<day>\d{2})-(?P<month>\d{2})-(?P<year>\d{4})"
text = "Date: 15-03-2023"
match = re.search(pattern, text)
if match:
    print(f"Day: {match.group('day')}")
    print(f"Month: {match.group('month')}")
    print(f"Year: {match.group('year')}")
```

#### Backreferences

```python
import re

pattern = r"(\w+) \1"
text = "hello hello world"
match = re.search(pattern, text)
print(match.group())
```

```python
import re

pattern = r"(?P<word>\w+) (?P=word)"
text = "hello hello world"
match = re.search(pattern, text)
print(match.group())
print(match.group('word'))
```

#### Positive Lookahead

```python
import re

pattern = r"\b\w+(?=ing\b)"
text = "I am running and jumping"
matches = re.findall(pattern, text)
print(matches)
```

#### Positive Lookbehind

```python
import re
pattern = r"(?<=\$)\d+"
text = "The price is $100 and 50 euros"
matches = re.findall(pattern, text)
print(matches)
```

#### Negative Lookbehind

```python
import re
pattern = r"(?<!\$)\b\d+"
text = "The price is $100 and 50 euros"
matches = re.findall(pattern, text)
print(matches)
```

#### Conditional Matching

```python
import re
pattern = r"(\d{3})(-)?(\s)?(\d{3})(?(2)-|\s)(\d{4})"
text1 = "123-456-7890"
text2 = "123 456 7890"
match1 = re.match(pattern, text1)
match2 = re.match(pattern, text2)
print(match1.group())
print(match2.group())
```

#### Using the | Operator for Alternation

```python
import re
pattern = r"cat|dog"
text = "I have a cat and a dog"
matches = re.findall(pattern, text)
print(matches)
```

```python
import re
pattern = r"I (love|hate) (cats|dogs)"
text = "I love cats and I hate dogs"
matches = re.findall(pattern, text)
print(matches)
```

#### Ex2: Validating and Parsing Emails

```python
import re

# Originally code did not work
email_pattern = r"""
    (?P<username>[-\w.+]+)       # Username (added '+' to the character class in order to get the last email to parse
    @                            # @ symbol
    (?P<domain>[-\w]+)           # Domain name
    \.                           # Dot
    (?P<tld>[a-z]{2,63}          # Top-level domain
    (?:\.[a-z]{2,63})?)          # Optional second-level domain
"""

email_regex = re.compile(email_pattern, re.VERBOSE)

emails = [
    "john.doe@example.com",
    "jane_smith123@company.co.uk",
    "invalid.email@.com",
    "another@invalid",
    "valid+email@subdomain.example.org"]

for email in emails:
    match = email_regex.match(email)
    if match:
        print(f"Valid email: {email}")
        print(f"Username: {match.group('username')}")
        print(f"Domain: {match.group('domain')}")
        print(f"TLD: {match.group('tld')}")
    else:
        print(f"Invalid email: {email}")
    print()
```
